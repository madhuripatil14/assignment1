```python
#6 Convert "Python" to uppercase without using .upper().
text = "Python"
result = ""

for ch in text:
    result += chr(ord(ch) - 32)

print(result)
```

```python
#7 Remove spaces from "Data analyst"
text = "Data Analyst"
result = ""
for ch in text:
    if ch != " ":
        result += ch
print(result)
```

```python
#8 Count words in "I am a Data Analyst"
text = "I am a Data Analyst"
words = text.split()
print(len(words))
```

```python
#9 Find the longest word in "I am learning Python"
text = "I am learning Python"
words = text.split()
longest_word = ""
for word in words:
    if len(word) > len(longest_word):
        longest_word = word
print(longest_word)
```

```python
#10 Find duplicate charecters in "programming"
text = "Programming"
for ch in set (text):
    if text.count(ch)>1:
        print(ch)
```

```python
#11 Count uppercase and lowercase letters
text = "Data Analyst"
Upper = 0
Lower = 0
for ch in text:
    if ch.isupper():
        Upper += 1
    elif ch.islower():
        Lower += 1
print("Uppercase", Upper)
print("Lowercase", Lower)
```
